## § 1 · Your Python toolbox

Five tools, one job each:

| Tool | Think of it as | Used for |
|---|---|---|
| Plain Python | a calculator | sums, lists, yes/no checks |
| **pandas** (`pd`) | Ramesh's register | open, check, clean and add columns to tables |
| **NumPy** (`np`) | a scientific calculator | square roots, averages, curves |
| **Matplotlib** (`plt`) | the drawing board | dots, lines, pictures |
| **scikit-learn** (`sklearn`) | the model kit | ready-made models; split, scale and score |

## § 2 · Reading Python in five minutes
**Read it like this**

- `=`: store: the name on the left remembers the value on the right
- `[ ]`: a list of values, or pick a column by name
- `( )`: give something to a tool, like `print(…)`
- `.`: use a tool that belongs to something: `pd.read_csv`
- `#`: a note for people; Python skips it

````python
# A name on the left, a value on the right. Python now remembers that price means 42.
price = 42

# Square brackets hold a list of values.
weights = [220, 310, 180]

# print() shows an answer on the screen. * means multiply.
print(price * 2)
````

````python
# len() counts the items in a list. A tool always takes round brackets.
print(len(weights))
````

````python
# import brings in a library. as pd gives pandas a short nickname.
import pandas as pd

# The pattern library.tool(input): pandas reads Ramesh’s register file into a table
# named df.
df = pd.read_csv("mango_register.csv")

# A column name in square brackets picks that column. Python shows all 15 weights,
# numbered from 0.
df["weight"]
````

````python
# A dot chains one more tool onto the result: the average weight. The text after # is a
# note for people.
print(df["weight"].mean())  # average
````

**The maths.** Describe each mango with two numbers. Find the tagged mangoes **closest** to the new one and let them **vote**.

> **distance = √( (softness₁ − softness₂)² + (smell₁ − smell₂)² )**

1. Mango B: softness 3, smell 2
2. to T3 (3, 1): √(0² + 1²) = **1**
3. to T2 (1, 0): √(2² + 2²) = √8 ≈ **2.8**
4. 3 nearest: T1, T3, T5 → all ripe → **Ripe**

*Softness: 1 hard, 2 firm, 3 soft. Smell: 0 nil, 1 faint, 2 sweet.*

````python
# From the scikit-learn model kit, take the nearest neighbours model.
from sklearn.neighbors import KNeighborsClassifier

# The clues of Ramesh’s six tagged mangoes, T1 to T6: [softness, smell] for each.
X = [[3, 2], [1, 0], [3, 1], [1, 1], [2, 2], [2, 0]]

# The answers to “Ripe?”, in the same order.
y = ["Yes", "No", "Yes", "No", "Yes", "No"]

# Step 1 of the recipe: an empty model that will ask the 3 nearest mangoes.
model = KNeighborsClassifier(n_neighbors=3)

# Step 2: fit means learn. For nearest neighbours, learning just means remembering every
# example.
model.fit(X, y)
````

````python
# Distances to the 3 nearest tagged mangoes, then which ones they are. Python counts
# from 0, so 0 is T1 and 2 is T3.
print(model.kneighbors([[3, 2]]))
````

````python
# Step 3: the vote. The new mango gets the answer most of its 3 neighbours have.
print(model.predict([[3, 2]]))
````

## § 14 · Growing a decision tree in Python
*Behind the slide “A classifier can be a flowchart”.*

**The maths.** At each step the tree tries every question and keeps the one that leaves the **purest piles**: the fewest mistakes.

> **Gini = 1 − (share ripe)² − (share not ripe)²**

1. “Soft?” Yes pile: 7 ripe, 1 not ripe
2. 1 − (7/8)² − (1/8)² ≈ **0.22** (almost pure)
3. No pile: 0 ripe, 7 not → 1 − 0 − 1 = **0** (pure)
4. 0 = pure pile · 0.5 = a perfect mix

*Bonus maths. Counting mistakes gives the same idea in words.*

````python
# Get ready: load what this section needs (already in place if you ran the sections above).
import pandas as pd
df = pd.read_csv("mango_register.csv")
df["soft_score"] = df["softness"].map({"Hard": 1, "Firm": 2, "Soft": 3})
df["smell_score"] = df["smell"].map({"Nil": 0, "Faint": 1, "Sweet": 2})
````

````python
# The decision tree model.
from sklearn.tree import DecisionTreeClassifier

# A tool to print the tree as text.
from sklearn.tree import export_text

# A yes/no clue: 1 if the mango is yellow or orange, 0 if green.
yo = {"Green": 0, "Yellow": 1, "Orange": 1}

# Store it as a new column.
df["yellow_orange"] = df["colour"].map(yo)

# Three clues for the tree to choose from.
X = df[["soft_score", "smell_score", "yellow_orange"]]

# max_depth limits how many questions it may ask in a row. random_state=2 makes it break
# ties between equally good questions the same way every time.
model = DecisionTreeClassifier(max_depth=3, random_state=2)

# Learn: find the best question, then the next, like 20 Questions.
model.fit(X, df["ripe"])
````

````python
# The tree as text. soft_score <= 2.50 means “not soft”. Each |--- step goes one
# question deeper.
print(export_text(model, feature_names=list(X.columns)))
````

## § 18 · From market to plate, in nine lines of Python
*Behind the slide “From market to plate”.*

**The kitchen, line by line**

- Buy: lines 1–4: bring in the tools and read the data (pandas)
- Wash: line 4: `drop_duplicates()` removes repeated rows
- Chop: line 5: choose the clues X and the answer y
- Hide: line 6: keep an exam pile aside (scikit-learn)
- Cook: lines 7–8: train on the practice pile only
- Taste: line 9: score it on the hidden exam pile

````python
# Buy. pandas, to bring the data into a table.
import pandas as pd

# A scikit-learn tool to split the data into a practice pile and an exam pile.
from sklearn.model_selection import train_test_split

# The model we will cook with.
from sklearn.tree import DecisionTreeClassifier

# Buy and wash. Read 100 mangoes, then drop any repeated rows.
df = pd.read_csv("mangoes_100.csv").drop_duplicates()

# Chop. Three clues, one answer. A comma lets one line fill two names.
X, y = df[["softness", "smell", "days"]], df["ripe"]

# Hide the exam. 80 mangoes for practice (tr = train), 20 kept back for the exam (te =
# test).
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.2, random_state=1)

# Cook. Choose a tree with up to 3 questions.
model = DecisionTreeClassifier(max_depth=3, random_state=1)

# Learn from the practice pile only. The exam pile stays hidden.
model.fit(X_tr, y_tr)
````

````python
# Taste. The share of the 20 exam mangoes it gets right.
print(model.score(X_te, y_te))
````

## § 23 · Finding the outlier with quartiles
*Behind the slide “Outliers”.*

**The maths.** Sort the values. The **quartiles** mark a quarter and three quarters of the way up. Anything far above the top quarter is suspicious.

> **fence = Q3 + 1.5 × (Q3 − Q1)**

1. Q1 = 145.75, Q3 = 152.25
2. gap = 152.25 − 145.75 = 6.5
3. fence = 152.25 + 1.5 × 6.5 = **162**
4. 1520 is far above → check it

````python
# pandas.
import pandas as pd

# The twelve heights from the slide, in cm. Kabir’s is the third.
h = pd.Series([148, 152, 1520, 145, 151, 143,
               150, 146, 149, 155, 141, 153])

# The summary. 25% is Q1, 50% is the median, 75% is Q3. Look how far max is from the
# rest.
h.describe()
````

````python
# quantile() finds Q1 and Q3.
q1, q3 = h.quantile(0.25), h.quantile(0.75)

# The fence: one and a half gaps above Q3.
top = q3 + 1.5 * (q3 - q1)

# Heights above 162 cm need checking.
print(top)
````

````python
# A filter: keep only the values above the fence. Only Kabir’s entry, at position 2
# (Python counts from 0).
h[h > top]
````

## § 35 · The lazy test, in Python
*Behind the slide “When accuracy fools you”.*

**The maths.** A lazy model always gives the most common answer. Its accuracy is just the size of the biggest group.

> **lazy accuracy = biggest group ÷ total**

2. 99 ÷ 100 = **99%**
3. sick people found: **0 of 1**
4. always compare a model with the lazy score

````python
# A deliberately lazy model, built for comparisons.
from sklearn.dummy import DummyClassifier

# 99 healthy people and 1 sick. * repeats a list; + joins two lists.
y = ["healthy"] * 99 + ["sick"]

# The lazy model ignores clues, so any clue will do.
X = [[0]] * 100

# It will always answer with the most common label.
lazy = DummyClassifier(strategy="most_frequent")

# Learning = noticing that “healthy” is most common.
lazy.fit(X, y)
````

## § 39 · The improvement loop: five exams, not one
*Behind the slide “The improvement loop”.*

**The maths.** Split the data into 5 parts. Each part takes a turn as the exam; the other 4 are practice. Report the **average**.

> **score = (s₁ + s₂ + s₃ + s₄ + s₅) ÷ 5**

1. depth 2: 0.90, 0.90, 1.00, 0.90, 0.85

3. every mango is in the exam exactly once

````python
# pandas.
import pandas as pd

# The five-exams tool (cross-validation).
from sklearn.model_selection import cross_val_score

# The model to improve.
from sklearn.tree import DecisionTreeClassifier

# 100 mangoes.
df = pd.read_csv("mangoes_100.csv")

# Clues and answers.
X, y = df[["softness", "smell", "days"]], df["ripe"]

# A tree with a chosen depth. Change it to improve the model.
model = DecisionTreeClassifier(max_depth=2, random_state=1)

# cv=5: five rounds, each with a different exam pile.
scores = cross_val_score(model, X, y, cv=5)

# The five exam scores, then their average.
print(scores.round(2), scores.mean().round(2))
````

## § G4 · The water tanks in Python: k-means
*Behind the slide “Where should the water tanks go?”.*

**The maths.** Two steps, repeated: each house joins its **nearest** tank; each tank moves to the **average position** of its houses.

> **centre = (x₁ + x₂ + … + xₙ) ÷ n, and the same for y**

1. houses at (2, 4), (4, 6), (6, 5)
2. x: (2 + 4 + 6) ÷ 3 = 4
3. y: (4 + 6 + 5) ÷ 3 = 5
4. the tank moves to **(4, 5)**; repeat until nothing changes

````python
# pandas, for the houses.
import pandas as pd

# Matplotlib, to draw the village.
import matplotlib.pyplot as plt

# The k-means model. k is the number of groups (tanks).
from sklearn.cluster import KMeans

# The 30 houses on the board: their x and y positions in metres.
houses = pd.read_csv("village.csv")

# n_clusters is k. n_init=10 tries 10 starting places and keeps the best.
km = KMeans(n_clusters=3, n_init=10, random_state=0)

# Learn the groups and give each house its tank number. Note: no answer key (no y). That
# is unsupervised learning.
houses["tank"] = km.fit_predict(houses[["x", "y"]])

# Where the tanks go: the average position of each group.
print(km.cluster_centers_.round(1))
````

````python
# Each house coloured by its tank.
plt.scatter(houses["x"], houses["y"], c=houses["tank"]); plt.show()
````

## § G5 · Supervised or unsupervised: one word in the code
*Behind the slide “Which kind of learning?”.*

**The maths.** Supervised learning gets the clues **and** the answers. Unsupervised learning gets only the clues.

> **supervised: fit(X, y) · unsupervised: fit(X)**

1. X = the clues, for both
2. y = the answer key, supervised only
3. unsupervised groups come back as numbers 0, 1, …
4. a person decides what each group means

````python
# A supervised model.
from sklearn.neighbors import KNeighborsClassifier

# An unsupervised model.
from sklearn.cluster import KMeans

# Four fruits: [size, sweetness].
X = [[1, 1], [1, 2], [8, 8], [9, 8]]

# The answer key.
y = ["lemon", "lemon", "orange", "orange"]

# fit(X, y): learning with the answer key.
sup = KNeighborsClassifier(n_neighbors=1).fit(X, y)

# It names the new fruit, because it learned the names.
print(sup.predict([[2, 1]]))
````

````python
# fit(X): no answer key. Find 2 groups.
uns = KMeans(n_clusters=2, n_init=10, random_state=0).fit(X)

# Group numbers only. The computer found the groups but cannot name them.
print(uns.labels_)
````

## § F1 · The line learns, in Python
Python finds the best line in one go. The loop below it is the film itself: the same four steps, round after round, from the same flat line at ₹20.

````python
# Python’s answer
import pandas as pd
from sklearn.linear_model import LinearRegression
# The 15 mangoes from Ramesh's register (written by § 0).
df = pd.read_csv("mango_register.csv")
# Python finds the best m and c in one go.
model = LinearRegression().fit(df[["weight"]], df["price"])
print(model.coef_, model.intercept_)
````

````python
# The training loop, as in the film
import numpy as np
# size: how far each mango is from 270 g, in hundreds of grams
x = (df["weight"].to_numpy() - 270) / 100
y = df["price"].to_numpy()
# Start like the film: a flat line at 20 rupees, step size 0.3.
tilt, height, step = 0.0, 20.0, 0.3
for rnd in range(43):
    # Steps 1 and 2: predict, then measure every gap.
    gap = y - (tilt * x + height)
    # Step 3: the slopes of the error map (which way is downhill).
    d_height = -2 * gap.mean()
    d_tilt = -2 * (gap * x).mean()
    # Step 4: a small step downhill.
    height = height - step * d_height
    tilt = tilt - step * d_tilt
print("m =", round(tilt / 100, 4), " c =", round(height - 2.7 * tilt, 2))
````

## § F2 · The border learns, in Python
LogisticRegression gives the border’s three numbers. The loop is the film: score, S-curve, miss, step, including the small penalty Python adds (C = 1).

````python
# Python’s answer
from sklearn.linear_model import LogisticRegression
# The 26 mangoes: softness and smell from 0 to 10.
pts = pd.read_csv("mango_clues.csv")
model = LogisticRegression().fit(pts[["softness", "smell"]], pts["ripe"])
# a and b, then c
print(model.coef_, model.intercept_)
# The share of the 26 on the right side of the border.
print(model.score(pts[["softness", "smell"]], pts["ripe"]))
````

````python
# The training loop, as in the film
# Clues measured from the middle (5): the steps settle faster.
X = pts[["softness", "smell"]].to_numpy() - 5
truth = (pts["ripe"] == "Yes").to_numpy() * 1.0
# Start like the film: a poor border, step size 2.
a, b, c, step, n = 0.5, -0.5, 0.0, 2.0, len(truth)
for rnd in range(76):
    # Step 1: a score for every mango, then the S-curve turns it into a chance.
    chance = 1 / (1 + np.exp(-(a * X[:, 0] + b * X[:, 1] + c)))
    # Steps 2 and 3: how far off each chance is, and which way to move.
    miss = chance - truth
    # Step 4: a small step downhill. a / n and b / n are Python's penalty.
    a = a - step * ((miss * X[:, 0]).mean() + a / n)
    b = b - step * ((miss * X[:, 1]).mean() + b / n)
    c = c - step * miss.mean()
print("a =", round(a, 2), " b =", round(b, 2), " c =", round(c - 5 * a - 5 * b, 2))
````
